## The question we're asking

Imagine two students both solving the same picture puzzle: *"find the hidden crown in this drawing."* One finds it in five seconds. The other takes much longer, or gives up.

**Could you predict which student would succeed just by watching where their eyes moved — before they even clicked an answer?**

That's the question this research asks, but scaled up: instead of one puzzle, **130 students** each completed **13 different hidden-object picture puzzles**, while a machine called an *eye tracker* recorded exactly where their eyes pointed, moment by moment, for the whole session.

## Why this matters

If eye movements really do reveal something about how well someone is solving a problem, that's useful in a few ways:

- **For educators**, it could help spot students who are struggling *while they're still working* — not just after they hand in a wrong answer.
- **For cognitive science**, it tests a long-standing idea: that *efficient visual attention* (looking at the right things, quickly, and not getting distracted) is part of what makes someone a good problem-solver.
- **For anyone curious about the mind**, it's a neat demonstration that a huge amount of information about *how* we think is hidden in something as simple as where we look.

## The study, in plain terms

- **130 students** took part.
- Each one saw **13 pictures**, one at a time, and had to find something hidden in it — a crown, a hat, a fish, a ring, and so on.
- A **chin rest and camera-based eye tracker** recorded their gaze the entire time, without them having to do anything special.
- When a student thought they'd found the answer, they **clicked on it with the mouse**.
- Afterwards, we know two things for every student: (1) *how many puzzles they actually solved correctly*, and (2) *the complete record of everywhere their eyes looked* while trying.

## One of the real puzzles

The boxes are invisible to participants in real life; they exist only so the eye-tracking software knows which named "zone" a person's gaze falls into at any moment.
