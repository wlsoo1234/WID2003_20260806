## Splitting students into two groups

To compare "strong" vs. "struggling" problem-solvers, we needed a simple, fair way to split 130 students into two groups. We used each student's overall accuracy across all 13 puzzles, and split the group right down the middle at the **midpoint (median)** score:

- Scored **at or above the midpoint** → labelled **High-performing**
- Scored **below the midpoint** → labelled **Low-performing**

This gave **77 High-performing** and **53 Low-performing** students — not a perfect 50/50 split (a few students tied exactly at the midpoint and were counted as High), but reasonably balanced.

For every one of the 13 puzzles, we measured each student's **dwell ratio on the correct spot** — the share of their total looking-time that landed on the right hiding place, rather than everywhere else on the picture. ## Was this a real pattern, or just luck?

Any time you compare two groups of people, you'll see *some* difference just by chance — the real question is whether the difference is big and consistent enough that it's unlikely to be a coincidence.

To check this rigorously, every measurement (dwell ratio, number of fixations, how quickly someone looked at the right spot, and so on — 267 measurements in total) was run through a statistical test that asks exactly that question. Rather than trusting eyeballing a chart, this test gives each measurement a "how surprising would this be by pure chance?" score.

The result: for the single most important measurement — **dwell ratio on the correct spot** — the test came back so far below the threshold for "just chance" that it's effectively guaranteed to be a real, reproducible pattern, not a fluke. Here's the actual computed result for that top measurement, straight from the underlying data:

```python
# Run this cell to see the statistical test result for the single strongest measurement
import pandas as pd

results = pd.read_csv('../outputs/reports/05_mann_whitney_results.csv')
top = results.iloc[0]
print(f"Measurement:                {top['feature']}")
print(f"Average for High performers: {top['high_mean']:.3f}")
print(f"Average for Low performers:  {top['low_mean']:.3f}")
print(f"How surprising would this be by pure chance? {top['p_fdr']:.2e}  (extremely unlikely — this is a real pattern)")
```

In plain terms: High-performing students, on average, spent about **21%** of their time looking at the correct spot, versus about **14%** for Low-performing students — and this gap is far too consistent across 130 students to be explained by chance alone.
