## Background, Explained Simply

### Why look before modelling?

A few things can go wrong if you skip straight to building a predictive model: it might accidentally latch onto an irrelevant coincidence in the data and look impressively "accurate" for the wrong reasons, or the data might have obvious quality problems that need fixing first. Looking carefully at the raw comparisons first is how you catch both.

### Reading a "violin" chart

A violin chart shows the *whole shape* of how a measurement is distributed across many people — not just one average number, but where most people cluster, and where the wide/narrow parts are. It's called a violin chart because the shape often resembles one. Wider sections mean more people had that particular value.

### Was the difference real, or just luck?

Comparing two groups of people, you'll always see *some* difference by pure chance — the question is whether it's big and consistent enough to trust. This study used a statistical test (specifically, the **Mann-Whitney U test**, which works well for the kind of lopsided, right-skewed numbers common in eye-tracking data) on every single measurement, and asked: *"if there were truly no real difference between High and Low performers, how surprising would a gap this large be?"* A very small answer means the pattern is very unlikely to be a fluke.

### The "testing lots of things at once" problem

Here's a subtlety worth knowing: if you test 100 different measurements at a relaxed threshold, you'd expect roughly 5 of them to look "significant" purely by chance, even if none of them are real. Testing many things at once inflates your odds of a false alarm. To correct for this, a technique called **FDR correction** (controlling the "false discovery rate") was applied — it tightens the bar in proportion to how many things were tested, so the final list of "real" patterns is much more trustworthy.

### Comparing search paths

A **scanpath** is the trail your eyes leave behind while searching — literally the path connecting each pause (fixation) to the next, like footprints. Comparing two students' scanpaths side by side is one of the most intuitive ways to *see* a difference in search strategy, beyond just numbers.

## What Actually Happened, By the Numbers

- **416 measurements** were tested. **136** looked significant before correcting for multiple testing; after the stricter FDR correction, **59** remained — still a substantial, trustworthy set of real patterns.
- The single strongest pattern: **dwell ratio on the correct spot**. High-performing students spent about **21%** of their time there on average, versus about **14%** for Low-performing students — and this gap was so consistent it would be expected to happen by pure chance roughly **1 in 2 million** times (technically: p ≈ 0.0000006 after correction). This is about as close to certain as behavioural research gets.
- The second-strongest pattern: **average fixation duration** — High performers' individual pauses lasted about **514 milliseconds** on average, versus **379 milliseconds** for Low performers — a real, meaningfully different pace of looking.
